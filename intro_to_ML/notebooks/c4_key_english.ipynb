## "Here's the takeaway:"
**Takeaway** es una de esas palabras comodín en inglés técnico, de negocios y educativo. Literalmente significa:\
\
"lo que te llevas para llevar".\
\
-> (La moraleja, la conclusión clave, la enseñanza principal o el resumen definitivo.)


- **either:** Se usa para indicar "uno u otro" de los dos elementos que se van a mencionar a continuación

- **Spurious patterns:** Patrones espurios. En estadística y Machine Learning, algo espurio significa que parece un patrón real o una causa, pero en realidad es falso, coincidencia pura o "ruido"

- **Recur (del inglés to recur):** Volver a suceder, repetirse, retornar o reaparecer en el tiempo. \
\
En el texto se habla de que el overfitting toma patrones "casualidad" que no se volveran a repetir (won't recur), Por ejemplo, una casa vendida solo porque el dueño pintó la puerta de verde.

- **Leading to (del verbo to lead to):** Conducir a, llevar a, o resultar en. Indica una relación causa y efecto directa.\
\
e. g. "...capturing spurious patterns..., leading to less accurate predictions".

- **Failing to (del verbo to fail to):** No lograr, fallar en, o dejar de hacer algo.\
\
e. g. "...failing to capture relevant patterns" -> no lograr capturar los patrones relevantes / fracasar en capturar... No implica fracaso dramatico si no más bien "no lograr" o **"no conseguir"**.\
\
En el underfitting, el modelo es tan simple que falla en (o es incapaz de) detectar las tendencias importantes de los datos.
