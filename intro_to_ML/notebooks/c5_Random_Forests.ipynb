# Class: Random Forests
---

Many other models are sensitive to getting the right parameters (sensibles a acertar con los parámetros correctos).

Significa que esos modelos (los otros) son muy delicados o estrictos. Si no le atinas exactamente a los valores ideales de sus hiperparámetros, el rendimiento del modelo se viene abajo de golpe. *Un Random Forest suele ser muy tolerante*.

```python
import pandas as pd
    
# Load data
melbourne_file_path = '../input/melbourne-housing-snapshot/melb_data.csv'
melbourne_data = pd.read_csv(melbourne_file_path) 
# Filter rows with missing values
melbourne_data = melbourne_data.dropna(axis=0)
# Choose target and features
y = melbourne_data.Price
melbourne_features = ['Rooms', 'Bathroom', 'Landsize', 'BuildingArea', 
                        'YearBuilt', 'Lattitude', 'Longtitude']
X = melbourne_data[melbourne_features]

from sklearn.model_selection import train_test_split

# split data into training and validation data, for both features and target
# The split is based on a random number generator. Supplying a numeric value to
# the random_state argument guarantees we get the same split every time we
# run this script.
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state = 0)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

forest_model = RandomForestRegressor(random_state=1)
forest_model.fit(train_X, train_y)
melb_preds = forest_model.predict(val_X)
print(mean_absolute_error(val_y, melb_preds))
```

## Conclusion
---
The random forests have parameters (or better said, "Hyperparameter") that we can change to improve their performance. But one of the best features of Random Forest models is that they generally work reasonably well even without this tuning. 

### About Random Forests' Sensitive
---


- Modelos robustos (poco sensibles): Un Random Forest suele ser muy tolerante. Si te equivocas un poco con la profundidad (max_depth) o la cantidad de árboles (n_estimators), el modelo aun así suele dar resultados decentes. No es tan sensible.

- Modelos frágiles (muy sensibles): Otros modelos más complejos o ciertos algoritmos de Deep Learning requieren que afines los hiperparámetros con precisión milimétrica (tasas de aprendizaje exactas, regularización muy específica, etc.). Si te pasas por un decimal, el modelo empieza a fallar estrepitosamente o hace overfitting. Eso significa que son sensibles a elegir los parámetros correctos.
\
\
En resumen: la frase te advierte que seguir entrenando modelos más avanzados está bien, pero muchos de ellos te van a exigir un trabajo mucho más duro de ajuste (tuning) porque no perdonan un mal parámetro.
